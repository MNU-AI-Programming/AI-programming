## 실행 순서가 중요합니다

- 위 칸을 건너뛰면 아래 칸에서 오류가 납니다.
- 순서가 꼬였으면 위에서부터 다시 실행합니다.
- **[런타임] → [모두 실행]** 이 가장 확실합니다.
- 오류가 나면 먼저 이것부터 해 보세요.

## 런타임이 무엇인가요?

- 내 코드를 대신 돌려 주는 구글의 컴퓨터입니다.
- 한참 안 쓰면 연결이 끊깁니다.
- 끊기면 위에서부터 다시 실행하면 됩니다.
- 노트북 파일 자체가 사라지는 것은 아닙니다.

> 📖 **런타임(runtime)** — 코드를 실제로 실행해 주는 임시 컴퓨터. 쉬면 꺼지고 다시 켜집니다.

## 무엇이 남고 무엇이 사라지나요?

| 구분 | 어디에 있나 | 세션이 끝나면 |
|---|---|---|
| 노트북 파일(.ipynb) | 구글 드라이브 | **남습니다** |
| 코드 · 설명 · 결과 | 노트북 안 | **남습니다** |
| `/content` 에 만든 파일 | 런타임 안 | 사라질 수 있습니다 |
| 그때 설치한 것 | 런타임 안 | 사라질 수 있습니다 |

파일 저장은 9주차에 자세히 배웁니다.

## 내 노트북은 어디에 저장되나요?

- 따로 누르지 않아도 구글 드라이브에 저장됩니다.
- 드라이브의 **'Colab Notebooks'** 폴더에 있습니다.
- 못 찾으면 **[파일] → [노트 열기]** 로 찾습니다.
- 제목을 잘 붙여 두면 나중에 찾기 쉽습니다.

## 저장하고 제출하기

1. `Ctrl + S` 로 한 번 저장합니다.
2. 오른쪽 위 **[공유]** 를 누릅니다.
3. 링크 권한을 **'링크가 있는 모든 사용자(뷰어)'** 로 바꿉니다.
4. 링크를 복사해 제출합니다.

**또는:** [파일] → `.ipynb` 다운로드 → 학교 LMS 업로드
