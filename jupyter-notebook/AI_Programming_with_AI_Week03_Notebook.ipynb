## 변수가 무엇인가요?

- 값을 넣어 두고 이름으로 다시 꺼내 쓰는 것입니다.
- 같은 값을 여러 번 쓸 때 훨씬 편합니다.
- 값이 바뀌면 한 곳만 고치면 됩니다.

> 📖 **변수(variable)** — 값에 이름표를 붙여 보관하는 상자. 택배 상자에 '겨울옷'이라 써 붙이는 것과 같습니다.


## 변수에 값 넣고 꺼내기

- =는 '같다'가 아니라 '넣어라'입니다.
- print()는 값을 화면에 보여 줍니다.
- 이름만 쓰면 안에 든 값이 나옵니다.

**실행 결과:**
```
김민수
20
```

> 📖 **할당(assignment)** — 변수에 값을 넣는 일. 상자에 물건을 담는 것과 같습니다.


````python
name = "김민수"
age = 20
print(name)
print(age)
````

## 변수 이름 짓는 법

- 영문, 숫자, 밑줄(_)로 짓는 것이 약속입니다.
- 숫자로 시작할 수 없습니다. (2day 는 안 돼요)
- 뜻이 드러나는 이름이 좋습니다. (a 보다 score)
- 대문자와 소문자는 서로 다른 이름입니다.


## 자료형이 무엇인가요?

- 값에도 종류가 있습니다.
- 숫자 10과 글자 "10"은 다릅니다.
- 종류가 다르면 할 수 있는 일도 다릅니다.

> 📖 **자료형(data type)** — 값의 종류. 옷, 그릇, 책을 서로 다른 서랍에 나눠 넣는 것과 같습니다.


## 네 가지 자료형

- int — 정수. 예: 10, -3, 0
- float — 소수. 예: 3.14, -0.5
- str — 글자. 예: "안녕" (따옴표로 감쌉니다)
- bool — 참(True) 또는 거짓(False)


## 자료형 확인하기

- type()은 값의 종류를 알려 줍니다.
- class는 '종류'라는 뜻입니다.
- 10은 숫자, "10"은 글자입니다.
- 따옴표가 붙으면 글자입니다.

**실행 결과:**
```
<class 'int'>
<class 'str'>
```


````python
print(type(10))
print(type("10"))
````

## 글자와 숫자를 섞으면?

⚠️ 이 코드는 **일부러 에러를 내서** 에러 메시지를 어떻게 읽는지 보여 주는 예시입니다. 실행하면 에러가 나는 것이 정상입니다.

- "10"은 글자, 3은 숫자입니다.
- 종류가 다르면 더할 수 없습니다.
- 코랩엔 이 줄 위로 몇 줄 더 나옵니다.
- 맨 아래 이 줄만 보면 됩니다.

**실행 결과 (에러):**
```
TypeError: can only concatenate str (not "int") to str
```

> 📖 **에러(error)** — 프로그램이 멈추면서 이유를 알려 주는 신호. '이 길은 막혔어요' 표지판과 같습니다.


````python
print("10" + 3)
````

## 에러 메시지 읽는 법

- 빨간 글씨는 고장이 아니라 안내문입니다.
- 맨 아래 줄부터 읽으세요. 이유가 거기 있습니다.
- 위쪽의 line 1 처럼 적힌 숫자가 틀린 줄 번호입니다.
- 고친 뒤 다시 실행하면 됩니다.

> 📖 **디버깅(debugging)** — 틀린 곳을 찾아 고치는 일. 옷에 붙은 실밥을 하나씩 찾아 떼는 것과 같습니다.


## 자주 만나는 에러 세 가지

- SyntaxError — 문법이 틀렸어요.
  - 예: 따옴표나 괄호를 빠뜨렸을 때
- NameError — 그런 이름이 없어요.
  - 예: 이름을 잘못 적었을 때
- TypeError — 종류가 안 맞아요.
  - 예: 글자와 숫자를 더했을 때


## 글자를 숫자로 바꾸기

- 글자끼리 +는 이어 붙입니다.
- 숫자끼리 +는 더합니다.
- int()는 글자를 숫자로 바꿉니다.
- 103은 숫자가 아니라 글자입니다.

**실행 결과:**
```
103
13
```

> 📖 **형 변환(casting)** — 값의 종류를 바꾸는 일. 동전을 지폐로 바꾸는 것과 같습니다.


````python
print("10" + "3")
print(int("10") + 3)
````

## 입력받기

- input(): 네모 칸에 21을 치고 Enter.
- 받은 값은 언제나 글자입니다.
- int()로 감싸야 계산이 됩니다.
- 코랩에서는 입력한 21도 위에 보입니다.

**실행 결과:**
```
42
```


````python
a = int(input())
print(a * 2)
````

## 계산에 쓰는 기호

- + 더하기, - 빼기, * 곱하기, / 나누기
- // 는 몫만, % 는 나머지만 알려 줍니다.
- 괄호 안을 가장 먼저 계산합니다.
- 곱셈·나눗셈이 덧셈·뺄셈보다 먼저입니다.

> 📖 **연산자(operator)** — 계산을 시키는 기호. 계산기의 +, - 버튼과 같습니다.


## 몫과 나머지

- 17을 5로 나누면 몫 3, 나머지 2.
- //는 몫, %는 나머지입니다.
- 17 / 5는 3.4가 나옵니다.
- 짝수·홀수 구분에 자주 씁니다.

**실행 결과:**
```
3
2
```


````python
print(17 // 5)
print(17 % 5)
````

## 참인가 거짓인가

- ==는 '같은가?', !=는 '다른가?'를 묻습니다.
- > 와 < 는 '크다', '작다'를 묻습니다.
- 답은 True 아니면 False 하나뿐입니다.
- 다음 주에는 이 True/False로 '만약 ~라면'을 만듭니다.

> 📖 **불(bool)** — 참(True)과 거짓(False) 둘 중 하나인 값. 예/아니오 스위치와 같습니다.


## 보기 좋게 출력하기 (f-string)

- 따옴표 앞에 f를 붙입니다.
- { } 안의 이름이 값으로 바뀝니다.
- 문장과 값을 한 줄에 씁니다.
- f를 빠뜨리면 글자 그대로 나옵니다.

**실행 결과:**
```
점수: 92점
```

> 📖 **f-문자열(f-string)** — 문장 안에 값을 끼워 넣는 따옴표. 빈칸 채우기 문제와 같습니다.


````python
score = 92
print(f"점수: {score}점")
````

## With AI: 오늘은 '고치는 용도'로만

- AI가 대신 써 준 코드는 내 실력이 되지 않습니다.
- 중간고사 전까지 AI 사용 규칙은 딱 하나입니다.
- ① 되는 것 — 내가 쓴 코드의 에러 찾기와 고치기
- ② 안 되는 것 — 코드를 만들어 달라고 하기
- 지금 받아 쓰면 중간고사 뒤에 코드를 못 읽습니다.
- 만드는 데 쓰는 법은 중간고사 뒤에 배웁니다.


## 프롬프트 템플릿 (오류 수정)

```
① 내가 쓴 코드예요:
(여기에 코드 전체를 붙여넣기)
② 에러로 나온 빨간 글씨예요:
(한 줄도 빼지 말고 붙여넣기)
③ 이 에러가 왜 났는지 한국어로 알려 주세요.
④ 전체 코드는 주지 말고, 고친 곳만 알려 주세요.
```

- 코드는 일부 말고 전체를 넣습니다.
- 빨간 글씨는 한 줄도 빼지 마세요.
- '왜'를 꼭 같이 물어보세요.
- '코드 짜 줘'는 쓰지 않습니다.

> 📖 **프롬프트(prompt)** — AI에게 시키는 말. 무엇을 어떻게 해 달라고 적은 주문서.
