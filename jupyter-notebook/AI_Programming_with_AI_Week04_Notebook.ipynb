## 조건문이 필요한 이유

- 지금까지 코드는 위에서 아래로 한 줄씩 실행됐습니다.
- 그런데 현실은 상황에 따라 답이 달라집니다.
- 점수가 90점이면 A, 아니면 다른 등급입니다.
- 이 갈림길을 만드는 것이 조건문입니다.

## 조건이 맞을 때만 실행하기

- `if` 뒤의 조건이 맞으면 안쪽 줄을 실행합니다.
- 조건 줄 끝에는 콜론(`:`)을 붙입니다.
- 95는 90보다 크므로 A가 나옵니다.

**실행 결과:**
```
A
```

> 📖 **조건문(if)** — '만약 ~라면'으로 길이 갈리는 문장. 신호등 앞에서 갈지 멈출지 정하는 것과 같습니다.

````python
score = 95
if score >= 90:
    print("A")
````

## 들여쓰기가 묶음을 정합니다

- 조건 줄 다음에 오는 안쪽 줄들이 한 묶음입니다.
- 묶음은 앞에 빈칸 네 칸을 넣어 표시합니다.
- 빈칸을 빼먹으면 오류가 납니다.
- 빈칸 개수가 들쭉날쭉해도 오류가 납니다.

> 📖 **들여쓰기(indent)** — 줄 앞의 빈칸. 파이썬은 이것으로 어디까지가 한 묶음인지 판단합니다.

## 아니면 이렇게 (else)

- `else` 는 '위 조건이 아닐 때'입니다.
- `else` 뒤에는 조건을 쓰지 않습니다.
- 둘 중 하나만 실행됩니다.

**실행 결과:**
```
B
```

````python
score = 75
if score >= 90:
    print("A")
else:
    print("B")
````

## 갈래가 셋일 때 (elif)

- `elif` 는 '앞이 아니면 이건 어때?'입니다.
- 위에서부터 차례로 확인합니다.
- 맞는 것 하나를 만나면 거기서 끝냅니다.
- `elif` 는 여러 번 쓸 수 있습니다.

**실행 결과:**
```
B
```

````python
score = 85
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("C")
````

## 조건을 쓸 때 조심할 것

- `=` 는 '넣어라', `==` 는 '같은가?'입니다.
- 순서가 중요합니다. 큰 범위를 먼저 쓰면 안 됩니다.

## 반복문이 필요한 이유

- 같은 줄을 백 번 복사해 붙일 수는 없습니다.
- 되풀이할 규칙만 적으면 컴퓨터가 알아서 반복합니다.
- 횟수가 정해졌으면 `for` 를 씁니다.
- 조건이 맞는 동안이면 `while` 을 씁니다.

> 📖 **반복문(for/while)** — 같은 일을 여러 번 되풀이하는 문장. 노래 후렴을 반복해 부르는 것과 같습니다.

## 정해진 횟수만큼 (for)

- `range(3)` 은 0, 1, 2 세 개를 만듭니다.
- `i` 에 그 값이 하나씩 들어갑니다.
- 안쪽 줄이 세 번 실행됩니다.

**실행 결과:**
```
0
1
2
```

````python
for i in range(3):
    print(i)
````

## range 읽는 법

- `range(5)` — 0부터 4까지 다섯 개입니다.
- 0부터 시작하고, **끝 숫자는 포함하지 않습니다.**
- `range(1, 4)` — 1부터 3까지 세 개입니다.
- 끝 숫자가 빠지는 것만 기억하면 됩니다.

## 1부터 세고 싶을 때

- 시작 숫자를 앞에 적습니다.
- 끝 숫자 4는 나오지 않습니다.
- 1부터 3까지 세 번 실행됩니다.

**실행 결과:**
```
1
2
3
```

````python
for i in range(1, 4):
    print(i)
````

## 조건이 맞는 동안 (while)

- 몇 번 반복할지 미리 모를 때 씁니다.
- 조건이 맞는 동안 계속 되풀이합니다.
- 조건이 틀어지는 순간 반복을 끝냅니다.
- 그래서 **안쪽에서 조건을 바꿔 줘야** 합니다.

## 세어 내려가기 (while)

- `n` 이 0보다 큰 동안 되풀이합니다.
- `n = n - 1` 이 n을 줄여 줍니다.
- n이 0이 되면 조건이 틀어져 끝납니다.

**실행 결과:**
```
3
2
1
```

````python
n = 3
while n > 0:
    print(n)
    n = n - 1
````

## 멈추지 않는 반복을 조심하세요

- 안쪽에서 조건을 바꾸지 않으면 영원히 돕니다.
- 코랩에서는 칸 왼쪽 **■ 중지** 단추로 멈춥니다.
- `while` 을 쓸 때는 '무엇이 끝을 만드나'를 먼저 정합니다.
- 이것이 `while` 에서 가장 흔한 실수입니다.

> 📖 **무한 루프(infinite loop)** — 끝나지 않고 영원히 도는 반복. 끝내는 조건을 빠뜨리면 생깁니다.

## 차곡차곡 더하기

- 더할 값을 담을 이름을 0으로 시작합니다.
- 반복할 때마다 그 이름에 더해 줍니다.
- 1+2+3+4+5 = 15 가 나옵니다.
- '처음 값 → 반복하며 갱신'이 기본 꼴입니다.

**실행 결과:**
```
15
```

````python
total = 0
for i in range(1, 6):
    total = total + i
print(total)
````
