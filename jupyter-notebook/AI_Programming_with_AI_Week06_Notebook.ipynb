## 값이 여러 개일 때

- 학생 30명의 점수에 이름을 30개 붙일 수는 없습니다.
- 대괄호 `[ ]` 안에 쉼표로 나열하면 한 줄에 담깁니다.
- 담은 순서가 그대로 유지됩니다.
- 숫자와 글자를 섞어 담을 수도 있습니다.

> 📖 **리스트(list)** — 값 여러 개를 순서대로 담는 상자. 번호가 붙은 사물함 한 줄과 같습니다.

## 담고 꺼내기

- 대괄호 안 번호로 하나씩 꺼냅니다.
- 첫 번째는 1번이 아니라 **0번**입니다.
- `-1` 은 맨 뒤를 뜻합니다.

**실행 결과:**
```
10
30
```

> 📖 **인덱스(index)** — 리스트 안 자리 번호. 사물함에 붙은 번호표와 같습니다.

````python
nums = [10, 20, 30]
print(nums[0])
print(nums[-1])
````

## 번호 읽는 법

- 번호는 **0부터** 시작합니다.
- 값이 세 개면 번호는 0, 1, 2 입니다.
- 마지막은 `-1`, 그 앞은 `-2` 로도 셀 수 있습니다.
- 없는 번호를 부르면 오류가 납니다.

## 없는 번호를 부르면

⚠️ 이 칸은 **일부러 에러를 내는** 칸입니다.

- 값은 세 개인데 3번을 불렀습니다.
- 쓸 수 있는 번호는 0, 1, 2 뿐입니다.
- `out of range` 는 '범위 밖'이라는 뜻입니다.

**실행 결과 (에러):**
```
IndexError: list index out of range
```

````python
nums = [10, 20, 30]
print(nums[3])
````

## 값 바꾸기와 개수 세기

- 번호를 골라 값을 덮어쓸 수 있습니다.
- `len` 은 담긴 값이 몇 개인지 알려 줍니다.
- 값을 바꿔도 개수는 그대로입니다.

**실행 결과:**
```
[10, 99, 30]
3
```

````python
nums = [10, 20, 30]
nums[1] = 99
print(nums)
print(len(nums))
````

## 일부만 잘라 쓰기

- 콜론 앞은 시작 번호, 뒤는 끝 번호입니다.
- **끝 번호는 포함되지 않습니다.** `range` 와 같습니다.
- 앞을 비우면 처음부터라는 뜻입니다.
- 잘라 낸 결과는 새 리스트입니다.

**실행 결과:**
```
[1, 2, 3]
[0, 1]
```

````python
a = [0, 1, 2, 3, 4]
print(a[1:4])
print(a[:2])
````

## 뒤에 하나 붙이기

- 점을 찍고 `append` 를 부릅니다.
- 맨 뒤에 하나가 붙습니다.
- `nums` 자체가 바뀝니다. 돌려받을 필요가 없습니다.

**실행 결과:**
```
[10, 20, 30]
```

````python
nums = [10, 20]
nums.append(30)
print(nums)
````

## 맨 뒤에서 하나 빼기

- `pop` 은 빼면서 그 값을 **돌려줍니다.**
- 돌려받은 값을 이름에 담아 쓸 수 있습니다.
- 빠진 자리만큼 리스트가 짧아집니다.

**실행 결과:**
```
30
[10, 20]
```

````python
nums = [10, 20, 30]
x = nums.pop()
print(x)
print(nums)
````

## 값을 보고 지우기

- 번호가 아니라 **값**을 적어 지웁니다.
- 같은 값이 여럿이면 앞의 하나만 지웁니다.
- 없는 값을 지우려 하면 오류가 납니다.

**실행 결과:**
```
[10, 30]
```

````python
nums = [10, 20, 30]
nums.remove(20)
print(nums)
````

## 줄 세우는 두 가지 방법

- `nums.sort()` — **원본을** 직접 줄 세웁니다.
- `sorted(nums)` — 줄 세운 **새 리스트를** 돌려줍니다.
- 원본을 남겨야 하면 `sorted` 를 씁니다.
- `sort()` 는 돌려주는 값이 없습니다. 담으면 `None` 입니다.

> 📖 **정렬(sort)** — 값을 순서대로 줄 세우는 일. 키 순서로 줄을 세우는 것과 같습니다.

## 원본을 두고 줄 세우기

- `nums` 는 그대로 남아 있습니다.
- 줄 세운 결과는 `new` 에 따로 담깁니다.
- 숫자는 작은 것부터 줄을 섭니다.
- 큰 것부터 하려면 `reverse=True` 를 넣습니다.

**실행 결과:**
```
[3, 1, 10]
[1, 3, 10]
```

````python
nums = [3, 1, 10]
new = sorted(nums)
print(nums)
print(new)
````

## 하나씩 꺼내 쓰기

- `range` 없이 리스트를 바로 쓸 수 있습니다.
- 담긴 값이 `s` 에 하나씩 들어옵니다.
- 담긴 개수만큼 되풀이됩니다.

**실행 결과:**
```
80
90
70
```

````python
scores = [80, 90, 70]
for s in scores:
    print(s)
````

## 합계와 평균

- `sum` 은 담긴 값을 모두 더해 줍니다.
- 평균은 합계를 개수로 나눕니다.
- 직접 `for` 로 더해도 결과는 같습니다.
- 빈 리스트를 나누면 오류가 나니 주의하세요.

**실행 결과:**
```
240
80.0
```

````python
scores = [80, 90, 70]
print(sum(scores))
print(sum(scores) / len(scores))
````

## 이름을 더 붙여도 같은 것입니다

- `b` 는 새 리스트가 아니라 **같은 것의 다른 이름**입니다.
- `b` 를 바꿨는데 `a` 도 같이 바뀌었습니다.
- 오류가 안 나서 알아채기 어렵습니다.
- 따로 쓰려면 `b = a.copy()` 로 복사합니다.

**실행 결과:**
```
[1, 2, 3, 4]
```

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
````
