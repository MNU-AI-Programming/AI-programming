## 튜플 — 바꿀 수 없는 묶음

- 리스트와 비슷하지만 한 번 만들면 못 바꿉니다.
- 대괄호 `[ ]` 대신 소괄호 `( )` 를 씁니다.
- 좌표 `(10, 20)` 처럼 바뀌면 안 되는 값에 씁니다.
- 실수로 값을 바꾸는 사고를 막아 줍니다.

> 📖 **튜플(tuple)** — 값을 순서대로 묶되 바꿀 수 없게 만든 것. 붙여 놓은 좌석 번호표와 같습니다.

## 튜플 만들고 꺼내기

- 꺼내는 방법은 리스트와 똑같습니다.
- 번호는 여기서도 0부터 셉니다.
- `len` 으로 개수도 셀 수 있습니다.

**실행 결과:**
```
10
2
```

````python
point = (10, 20)
print(point[0])
print(len(point))
````

## 튜플은 바꿀 수 없습니다

⚠️ 이 칸은 **일부러 에러를 내는** 칸입니다.

- 꺼내기는 되지만 넣기는 안 됩니다.
- `item assignment` 는 '자리에 값 넣기'입니다.
- 바꿔야 한다면 리스트를 써야 합니다.

**실행 결과 (에러):**
```
TypeError: 'tuple' object does not support item assignment
```

````python
point = (10, 20)
point[0] = 99
````

## 한 번에 나눠 담기

- 왼쪽에 이름을 개수만큼 적으면 나뉘어 담깁니다.
- `point[0]`, `point[1]` 로 적는 것보다 읽기 좋습니다.
- 개수가 안 맞으면 오류가 납니다.

**실행 결과:**
```
10
20
```

````python
point = (10, 20)
x, y = point
print(x)
print(y)
````

## 딕셔너리 — 이름표를 붙여 담기

- 리스트는 0번, 1번처럼 **번호**로 찾습니다.
- 그런데 사람 **이름**으로 점수를 찾고 싶을 때가 있습니다.
- 그럴 때 이름표(키)와 값을 짝지어 담습니다.
- 중괄호 `{ }` 안에 `키: 값` 으로 적습니다.

> 📖 **딕셔너리(dict)** — 이름표와 값을 짝지어 담는 것. 사전에서 낱말로 뜻을 찾는 것과 같습니다.

## 딕셔너리 만들고 찾기

- 대괄호 안에 번호 대신 이름표를 적습니다.
- 이름표만 알면 몇 개가 담겨 있든 바로 찾습니다.
- 이름표는 겹치지 않게 하나씩만 둡니다.

**실행 결과:**
```
90
```

````python
score = {"김민수": 90, "이영희": 80}
print(score["김민수"])
````

## 없는 이름표를 찾으면

⚠️ 이 칸은 **일부러 에러를 내는** 칸입니다.

- 담아 두지 않은 이름표를 찾았습니다.
- `KeyError` 뒤에 찾으려던 이름표가 나옵니다.
- 이번 주 가장 자주 만나는 오류입니다.

**실행 결과 (에러):**
```
KeyError: '박철수'
```

````python
score = {"김민수": 90}
print(score["박철수"])
````

## 없어도 괜찮게 찾기

- `get` 은 없으면 오류 대신 정해 준 값을 줍니다.
- 뒤에 적은 `0` 이 '없을 때 쓸 값'입니다.
- 있을지 없을지 모를 때는 `get` 을 씁니다.

**실행 결과:**
```
0
```

````python
score = {"김민수": 90}
print(score.get("박철수", 0))
````

## 넣고 고치기

- 없는 이름표에 넣으면 **새로 추가**됩니다.
- 있는 이름표에 넣으면 **값이 바뀝니다.**
- 넣는 방법은 같고 결과만 다릅니다.
- 담은 순서대로 남아 있습니다.

**실행 결과:**
```
{'김민수': 95, '박철수': 70}
```

````python
score = {"김민수": 90}
score["박철수"] = 70
score["김민수"] = 95
print(score)
````

## 하나씩 꺼내 쓰기

- `items()` 는 이름표와 값을 짝으로 줍니다.
- 받을 이름을 둘 적어 나눠 담습니다.
- 튜플에서 배운 나눠 담기와 같은 방식입니다.

**실행 결과:**
```
김민수 90
이영희 80
```

````python
score = {"김민수": 90, "이영희": 80}
for name, s in score.items():
    print(name, s)
````

## 몇 번 나왔는지 세기

- 딕셔너리의 가장 유명한 쓰임입니다.
- 처음 보는 글자는 `get` 이 `0` 을 줍니다.
- 거기에 1 을 더해 다시 담습니다.
- 글자 대신 단어를 세면 단어 빈도가 됩니다.

**실행 결과:**
```
{'b': 1, 'a': 3, 'n': 2}
```

````python
text = "banana"
freq = {}
for ch in text:
    freq[ch] = freq.get(ch, 0) + 1
print(freq)
````

## 집합 — 겹치지 않게 담기

- 같은 값을 여러 번 넣어도 하나만 남습니다.
- 중복을 없앨 때 가장 빠른 방법입니다.
- **담은 순서는 지켜 주지 않습니다.**
- 중괄호 `{ }` 를 쓰지만 이름표는 없습니다.

> 📖 **집합(set)** — 겹치는 값을 하나로 모아 담는 것. 명단에서 중복 이름을 지운 것과 같습니다.

## 중복 없애고 들어 있는지 보기

- `set` 으로 바꾸면 겹치는 값이 사라집니다.
- 순서가 필요하면 `sorted` 로 줄을 세웁니다.
- `in` 으로 들어 있는지 바로 확인합니다.
- 많은 값 중에서 찾을 때 특히 빠릅니다.

**실행 결과:**
```
[1, 2, 3]
True
```

````python
nums = [1, 1, 2, 3, 3]
s = set(nums)
print(sorted(s))
print(2 in s)
````

## 무엇을 언제 쓸까요

| 담는 방법 | 순서 | 중복 | 바꾸기 | 언제 쓰나 |
|---|---|---|---|---|
| 리스트 `[ ]` | 있음 | 됨 | 됨 | 순서가 있고 바꿔야 할 때 |
| 튜플 `( )` | 있음 | 됨 | **안 됨** | 바뀌면 안 될 때 |
| 딕셔너리 `{키: 값}` | 담은 순서 | 키는 하나씩 | 됨 | 이름표로 찾아야 할 때 |
| 집합 `{ }` | **없음** | **안 됨** | 됨 | 겹치면 안 되거나 있는지만 볼 때 |

고를 때 질문 — **바꿔야 하나? 이름표로 찾나? 겹쳐도 되나?**
