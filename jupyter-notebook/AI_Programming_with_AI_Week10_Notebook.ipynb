## 열과 행

- 열(column)은 세로로 늘어선 같은 종류 값.
- 행(row)은 가로로 한 줄, 한 건의 기록.
- 엑셀의 세로줄, 가로줄과 같습니다.
- 표를 볼 때는 열 이름부터 확인하세요.

> 📖 **행(row)** — 가로로 한 줄. 출석부에서 한 사람 줄을 왼쪽에서 오른쪽으로 읽는 것과 같습니다.


## 열 하나 꺼내기

- df["score"]로 그 열만 꺼냅니다.
- tolist()는 파이썬 목록으로 바꿔 줍니다.
- 대괄호 안 이름은 따옴표로 감쌉니다.

## 열 여러 개 꺼내기

## 빠진 값이 있으면

- 설문에서 답을 안 한 칸과 같습니다.
- 그냥 두면 평균 같은 계산이 틀어집니다.
- 먼저 몇 개나 비었는지부터 셉니다.
- 그다음 채우거나, 그 줄을 뺍니다.

> 📖 **결측치(missing value)** — 표에서 비어 있는 칸. 설문지에서 답을 안 적은 빈칸과 같습니다.


## 빠진 값 세어 보기

- isna()는 빈 칸이면 True를 매깁니다.
- sum()으로 열마다 빈 칸 개수를 셉니다.
- score 열에 하나 비어 있네요.

**실행 결과:**
```
name     0
score    1
dtype: int64
```


````python
import pandas as pd
df = pd.DataFrame({"name": ["가", "나", "다"],
                   "score": [90, None, 70]})
print(df.isna().sum())
````

## 빠진 값 채우거나 버리기

- fillna(0)은 빈 칸을 0으로 채웁니다.
- 0 대신 평균으로 채우기도 합니다.
- dropna()는 그 줄을 통째로 뺍니다.
- 빈 칸이 있었던 열은 숫자가 소수로 바뀝니다.

**실행 결과:**
```
[90.0, 0.0, 70.0]
```


````python
import pandas as pd
df = pd.DataFrame({"name": ["가", "나", "다"],
                   "score": [90, None, 70]})
print(df["score"].fillna(0).tolist())
````

## 평균과 최고점

- mean()은 평균, max()는 최댓값.
- 괄호를 빠뜨리면 숫자 대신 이상한 글자가 나옵니다.
- 빠진 값은 채운 뒤에 계산하세요.

**실행 결과:**
```
80.0
90
```


````python
import pandas as pd
df = pd.DataFrame({"name": ["가", "나", "다"],
                   "score": [90, 80, 70]})
print(df["score"].mean())
print(df["score"].max())
````

## 몇 명씩인지 세기

- value_counts()는 같은 값끼리 세어 줍니다.
- 원래는 표(Series) 모양으로 나옵니다.
- to_dict()로 보기 쉽게 정리했습니다.
- 설문 응답 집계에 자주 씁니다.

**실행 결과:**
```
{'상': 2, '중': 1, '하': 1}
```


````python
import pandas as pd
df = pd.DataFrame({"name": ["가", "나", "다", "라"],
                   "grade": ["상", "중", "상", "하"]})
print(df["grade"].value_counts().to_dict())
````

## NumPy는 무엇이 다른가요?

- 숫자만 담는, 계산이 빠른 목록입니다.
- 목록 전체를 한 번에 계산합니다.
- pandas 도 속은 NumPy로 되어 있습니다.

> 📖 **배열(array)** — 숫자만 순서대로 담은 상자. 계란판처럼 칸마다 숫자 하나씩 들어갑니다.


## 배열로 한 번에 계산하기

- 배열은 한 줄로 전체에 곱합니다.
- 파이썬 목록에 * 10을 하면 목록이 열 번 이어 붙습니다.
- 하나씩 곱하지 않아도 됩니다.

**실행 결과:**
```
[10 20 30]
```


````python
import numpy as np
a = np.array([1, 2, 3])
print(a * 10)
````
