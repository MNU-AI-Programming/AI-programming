## 가장 단순한 그래프

- `plt` 라는 짧은 이름으로 가져옵니다.

- 2) **꾸미기** — `title`, `xlabel`, `ylabel`

## 선그래프 — 시간에 따른 변화

- 순서대로 이어지는 값에 씁니다.
- `marker="o"` 를 주면 점이 찍힙니다.
- `grid(True)` 는 눈금선을 켭니다.
- 목요일이 가장 높다는 것이 바로 보입니다.

````python
import matplotlib.pyplot as plt

days = ["월", "화", "수", "목", "금"]
steps = [5200, 7400, 6800, 9100, 8000]

plt.plot(days, steps, marker="o")
plt.title("요일별 걸음 수")
plt.xlabel("요일")
plt.ylabel("걸음 수")
plt.grid(True)
plt.show()
````

## 막대그래프 — 무엇이 더 큰가

- 서로 다른 무리를 비교할 때 씁니다.
- 이름 목록과 값 목록의 **개수가 같아야** 합니다.
- 막대는 0에서 시작해야 비교가 정직합니다.
- 이름이 길면 겹치니 짧게 씁니다.

````python
import matplotlib.pyplot as plt

labels = ["1반", "2반", "3반"]
values = [78, 85, 71]

plt.bar(labels, values)
plt.title("반별 평균 점수")
plt.xlabel("반")
plt.ylabel("평균 점수")
plt.show()
````

## 히스토그램 — 어떻게 퍼져 있나

- 값 하나하나가 아니라 **퍼진 모양**을 봅니다.
- `bins` 는 구간을 몇 칸으로 나눌지입니다.
- 칸 수를 바꾸면 모양이 달라 보입니다.
- 막대그래프와 달리 값을 구간으로 묶습니다.

````python
import matplotlib.pyplot as plt

scores = [70, 85, 90, 60, 75, 88, 95,
          72, 68, 81, 77, 93, 84, 66]

plt.hist(scores, bins=5)
plt.title("점수 분포")
plt.xlabel("점수")
plt.ylabel("사람 수")
plt.show()
````

## 산점도 — 둘 사이 관계

- 점 하나가 사람 한 명입니다.
- 오른쪽 위로 올라가면 같이 커지는 관계입니다.
- **함께 움직인다고 원인은 아닙니다.**
- 멀리 떨어진 점은 따로 확인해 봅니다.

````python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
scores = [55, 60, 72, 78, 88, 95]

plt.scatter(hours, scores)
plt.title("공부 시간과 점수")
plt.xlabel("공부 시간")
plt.ylabel("점수")
plt.show()
````

## 어떤 그래프를 고를까요

| 보고 싶은 것 | 그래프 |
|---|---|
| 시간에 따른 변화 | 선그래프 `plot` |
| 무리끼리 비교 | 막대그래프 `bar` |
| 퍼진 모양 | 히스토그램 `hist` |
| 둘 사이 관계 | 산점도 `scatter` |

먼저 **'무엇이 궁금한가'** 를 정하면 답이 나옵니다.

## 선 두 개와 범례

- `plot` 을 여러 번 부르면 겹쳐 그려집니다.
- 각각에 `label` 로 이름을 붙입니다.
- `legend()` 를 불러야 이름표가 나옵니다.
- `label` 만 쓰고 `legend` 를 빼먹기 쉽습니다.

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]
a = [10, 20, 15]
b = [8, 18, 25]

plt.plot(x, a, marker="o", label="1반")
plt.plot(x, b, marker="o", label="2반")
plt.title("반별 점수 변화")
plt.legend()
plt.show()
````

## 그림 파일로 저장하기

- 과제에 그림을 넣을 때 씁니다.

- 순서가 바뀌면 빈 그림이 저장됩니다.

## 표 데이터로 바로 그리기

- 열 하나를 꺼내 바로 `plot` 을 부를 수 있습니다.
- `kind` 로 그래프 종류를 고릅니다.
- 꾸미기는 그대로 `plt` 로 이어서 합니다.
- 실제 파일은 `read_csv` 로 읽어 오면 됩니다.

> 📖 **데이터프레임(DataFrame)** — 행과 열로 된 표. 10주차에 배운 그것입니다.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({"score": [70, 85, 90,
                             60, 75, 88, 95]})

df["score"].plot(kind="hist", bins=5)
plt.title("점수 분포")
plt.xlabel("점수")
plt.show()
````
