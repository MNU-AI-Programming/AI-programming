## 분석은 '고르기'에서 시작합니다

- 표 전체를 한 번에 보는 일은 거의 없습니다.
- '80점 넘는 사람만' 처럼 조건을 걸어 봅니다.
- 고른 다음에 줄을 세우고 요약합니다.
- 이 세 걸음이 데이터 분석의 기본 흐름입니다.

> 📖 **필터링(filtering)** — 조건에 맞는 것만 골라내는 일. 체로 쳐서 큰 알갱이만 남기는 것과 같습니다.

## 조건으로 골라내기

- 대괄호 안에 조건을 그대로 씁니다.
- 조건에 맞는 줄만 남습니다.
- 왼쪽 숫자는 **원래 몇 번째 줄이었는지**입니다.
- 번호가 0, 1, 2 로 다시 매겨지지 않습니다.

**실행 결과:**
```
  name  score
1    나     85
2    다     90
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df[df["score"] >= 80])
````

## 조건은 참·거짓 목록입니다

- `df["score"] >= 80` 은 값 하나가 아닙니다.
- 줄마다 참인지 거짓인지 답한 **목록**입니다.
- 그 목록을 대괄호에 넣으면 참인 줄만 남습니다.
- 아래 칸으로 조건 자체를 출력해 보세요.

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df["score"] >= 80)
````

## 조건 두 개를 이어 붙일 때

- 둘 다 만족 — `&` 를 씁니다.
- 둘 중 하나라도 — `|` 를 씁니다.
- **각 조건을 반드시 괄호로 묶습니다.**
- `and` 와 `or` 는 여기서 쓸 수 없습니다.

**실행 결과:**
```
  name  score
1    나     85
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
cond = ((df["score"] >= 80)
        & (df["score"] < 90))
print(df[cond])
````

## and 를 쓰면 오류가 납니다

⚠️ 이 칸은 **일부러 에러를 내는** 칸입니다.

- `and` 는 참·거짓 **하나끼리** 쓰는 말입니다.
- 여기서는 참·거짓이 줄 개수만큼 있습니다.
- `ambiguous` 는 '어느 쪽인지 알 수 없다'입니다.
- `and` 를 `&` 로 바꾸고 괄호를 씌우면 됩니다.

**실행 결과 (에러):**
```
ValueError: The truth value of a Series is ambiguous.
Use a.empty, a.bool(), a.item(), a.any() or a.all().
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df[df["score"] >= 80
         and df["score"] < 90])
````

## 목록 안에 있는 것만

- 고를 값이 여러 개면 `isin` 이 편합니다.
- `==` 를 여러 번 이어 붙일 필요가 없습니다.
- 고를 값들을 대괄호 목록으로 넘깁니다.

**실행 결과:**
```
  name  score
0    가     70
2    다     90
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df[df["name"].isin(["가", "다"])])
````

## 범위로 고르기

- `between` 은 **시작과 끝을 모두 포함**합니다.
- `range` 나 슬라이싱과 다른 점입니다.
- `&` 를 쓴 것과 결과가 같지만 더 짧습니다.

**실행 결과:**
```
  name  score
1    나     85
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df[df["score"].between(80, 89)])
````

## 줄 세우기

- `sort_values` 에 기준 열 이름을 줍니다.
- 그냥 두면 작은 값부터 올라갑니다.
- `ascending=False` 를 주면 큰 값부터입니다.
- 원본은 그대로 있고 새 표가 나옵니다.

> 📖 **정렬(sort)** — 값을 순서대로 줄 세우는 일. 키 순서로 줄을 세우는 것과 같습니다.

**실행 결과:**
```
  name  score
2    다     90
1    나     85
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df.sort_values("score",
                     ascending=False).head(2))
````

## 위에서 몇 개만 바로 뽑기

- 줄 세우고 자르는 일을 한 번에 합니다.
- 작은 쪽은 `nsmallest` 를 씁니다.
- 몇 개, 어느 열 기준인지 둘 다 적습니다.

**실행 결과:**
```
  name  score
2    다     90
1    나     85
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df.nlargest(2, "score"))
````

## 골라내고 줄 세우고 뽑기

- 골라낸 결과에 이름을 붙여 두었습니다.
- 그 이름으로 이어서 줄을 세웁니다.
- 한 줄로 이어 써도 되지만 나누면 확인이 쉽습니다.
- 중간에 출력해 보면 어디가 틀렸는지 보입니다.

**실행 결과:**
```
  name  score
2    다     90
1    나     85
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
high = df[df["score"] >= 80]
print(high.sort_values("score",
                       ascending=False))
````

## 원하는 열만 함께 고르기

- `loc` 은 줄 조건과 열 이름을 함께 받습니다.
- **쉼표 앞이 줄, 뒤가 열**입니다.
- 필요한 열만 남기면 보기 편합니다.
- 열 이름은 대괄호 목록으로 적습니다.

**실행 결과:**
```
  name
1    나
2    다
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(df.loc[df["score"] >= 80, ["name"]])
````

## 골라내기 → 줄 세우기 → 요약

- 1) 조건으로 볼 대상을 좁힙니다.
- 2) 기준을 정해 줄을 세웁니다.
- 3) 몇 개인지, 평균이 얼마인지 요약합니다.
- 4) 요약을 **한 문장으로** 적습니다.

숫자만 뽑고 끝내면 분석이 아닙니다.

## 골라낸 것 요약하기

- `len` 은 몇 줄인지 세어 줍니다.
- `mean` 은 평균을 냅니다.
- `round` 로 소수점 자리를 줄입니다.
- **'80점 이상은 2명, 평균 87.5점'** 이 한 문장 요약입니다.

**실행 결과:**
```
2
87.5
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
high = df[df["score"] >= 80]
print(len(high))
print(round(high["score"].mean(), 1))
````
