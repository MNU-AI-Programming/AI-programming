## 연습 문제 — 이론

**Q) 두 조건을 함께 걸 때 맞는 것은?**

- A) `df[(a > 0) and (b > 0)]`
- B) `df[(a > 0) & (b > 0)]`
- C) `df[a > 0 & b > 0]`

✅ 내 답:

<details><summary>정답 보기</summary>

**B** — `&` 를 쓰고 각 조건을 괄호로 묶습니다.
A 는 `ValueError`, C 는 괄호가 없어 엉뚱하게 계산됩니다.

## 연습 실기 1. 평균과 최고점

- `mean()` 으로 평균을 냅니다.
- `round` 로 소수점 자리를 줄입니다.
- 최고점은 `nlargest(1, 열이름)` 이 편합니다.

**실행 결과:**
```
81.7
  name  score
2    다     90
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
print(round(df["score"].mean(), 1))
print(df.nlargest(1, "score"))
````

## 연습 실기 2. 골라내고 상위 두 명

- 먼저 조건으로 골라내고 이름을 붙입니다.
- 그다음 위에서 몇 개를 뽑습니다.
- `sort_values` 뒤에 `head(2)` 를 붙여도 같습니다.

**실행 결과:**
```
  name  score
2    다     90
1    나     85
```

````python
import pandas as pd
df = pd.DataFrame({
    "name": ["가", "나", "다"],
    "score": [70, 85, 90]})
high = df[df["score"] >= 80]
print(high.nlargest(2, "score"))
````

## 연습 실기 4. JSON으로 저장하기

- `dump` 는 저장, `load` 는 불러오기입니다.
- `ensure_ascii=False` 가 없으면 한글이 깨져 보입니다.
- **저장만 하지 말고 다시 읽어 확인**합니다.

**실행 결과:**
```
{'avg': 82.5, 'top': '김민수'}
```

````python
import json

result = {"avg": 82.5, "top": "김민수"}
p = "report.json"

with open(p, "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False)

with open(p, encoding="utf-8") as f:
    print(json.load(f))
````

## 연습 실기 5. 쓰레드 기본 (13주 범위)

- `Thread(target=일)` → `start()` → `join()`.
- **끝내는 조건**이 있어야 스스로 멈춥니다.
- `Queue` 와 `Event` 는 시험 범위가 아닙니다.

**실행 결과:**
```
작업 중
메인 끝
```

````python
import threading

def work():
    print("작업 중")

t = threading.Thread(target=work)
t.start()
t.join()
print("메인 끝")
````
